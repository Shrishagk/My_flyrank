## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is primarily a ranking/scoring task. The goal is not only to classify pages as good or bad, but to assign each content page a priority score and rank pages so that the highest-priority pages can be reviewed first. A binary decline indicator can be used as an intermediate target for model development, but the final output is a ranked review queue.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

For the starter experiment, I will use trend_direction == "down" as a proxy target for whether a content page is showing a declining trend. This label comes from an observed field in the starter dataset, rather than from a future outcome.

For the final warehouse experiment, I intend to define a future-looking target using a historical feature window followed by a future outcome window. This will better match the real decision of identifying pages that may require review before the observed decline occurs.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@50-Because my output is a ranked queue. If a reviewer can examine 50 pages, I care about how many of those top 50 actually correspond to the target.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is one content page. Each dataframe row represents one content_id and contains the observed search, content, engagement, and trend signals associated with that page.

The starter slice contains 30,000 rows, so the current analysis operates at the content-page level rather than at the individual query or daily-event level.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can capture one or two obvious conditions, but refresh opportunity depends on multiple signals that can interact. Search demand, CTR, position, engagement, content characteristics, and observed trend can produce different combinations across pages. A scoring model can learn patterns across these signals and produce a more flexible ranking than a single threshold-based rule.

However, I will only claim that ML is useful if it improves the chosen ranking metric against a transparent baseline under honest validation.
